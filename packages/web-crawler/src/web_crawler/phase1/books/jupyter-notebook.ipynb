```python
from urllib.parse import urljoin

import requests
from bs4 import BeautifulSoup
```

```python
BASE_URL = "https://books.toscrape.com/"
```

# 1 - No description

```python
response = requests.get(BASE_URL)

soup = BeautifulSoup(response.text, "html.parser")

# Get categories
categories_url = []

for cat in soup.select(".nav-list ul a"):
    categories_url.append(cat["href"])
```

```python
books = []


def scrape_book_info(soup: BeautifulSoup):
    for row in soup.select(".product_pod"):
        books.append(
            {
                "title": row.select("a[title]")[0]["title"],
                "rating": row.select(".star-rating")[0]["class"][1],
                "price": row.select(".price_color")[0].text[1:],
                "currency": row.select(".price_color")[0].text[0],
            }
        )


for cat_url in categories_url:
    print(cat_url)
    while True:
        cat_response = requests.get(urljoin(response.url, cat_url))
        cat_response.encoding = "utf-8"

        soup_cat = BeautifulSoup(cat_response.text, "html.parser")
        scrape_book_info(soup_cat)

        if soup_cat.select(".next a[href]"):  # next button
            cat_url = urljoin(cat_url, soup_cat.select(".next a[href]")[0]["href"])
        else:
            break
```
